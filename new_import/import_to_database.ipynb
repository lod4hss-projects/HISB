```python
import pandas as pd
from sqlalchemy import create_engine, text
```

## Import CSV to Postgresql on Localhost

We import here the two census CSV tables to a localhost docker database 'basel_database'

```python
# Create the database connection URL
# Format: postgresql://user:password@host:port/database_name
db_url = "postgresql://postgres:secret@localhost:5432/basel_database"

# Create the SQLAlchemy engine
engine = create_engine(db_url)
```

### Import VS 1850

```python
f_add = '/home/francesco/kDrive/kleioLab/Projekte/HISB_hotz_gerhard/daten_20260722/VZ_1850_H_END_20190701_inkl_Koordinaten.csv'

df = pd.read_csv(f_add, sep='\t')
df.head()
```

```python
# Replace '(' and ')' with '_' in column names
df.columns = df.columns.str.replace(r'[()]', '_', regex=True)
df.columns = df.columns.str.replace(r'\s', '_', regex=True)
df.columns = df.columns.str.strip('_')
df.columns = df.columns.str.replace('__', '_')
df.columns = df.columns.str.replace('/', '_')
df.columns = df.columns.str.replace('-', '_')
df.columns = df.columns.str.replace('.', '')
df.columns = df.columns.str.lower()
df.iloc[:3, 10:20]
```

```python
print(df.columns)
```

```python

# Save the DataFrame to a new PostgreSQL table
# if_exists='fail' raises an error if the table exists (use 'replace' to overwrite or 'append' to add)
df.to_sql('vz_1850', engine, schema='hisb_import', if_exists='fail', index=False)
```

```python

with engine.connect() as conn:
     # conn.execute(text("ALTER TABLE hisb_import.vz_original ADD PRIMARY KEY (id_nr);"))
    conn.commit()

# Dispose the engine
engine.dispose()
```
